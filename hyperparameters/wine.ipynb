```python
# Load the wine datasets with the correct delimiter
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

red_wine = pd.read_csv("/home/darshchouhan/Downloads/wine+quality/winequality-red.csv", sep=';')
white_wine = pd.read_csv("/home/darshchouhan/Downloads/wine+quality/winequality-white.csv", sep=';')

# Combine for easier analysis
red_wine['type'] = 'red'
white_wine['type'] = 'white'
wine = pd.concat([red_wine, white_wine], ignore_index=True)

# 1. Data info and statistics
print("Red wine info")
red_wine.info()
print("\nWhite wine info")
white_wine.info()

print("\nRed wine describe()")
print(red_wine.describe())
print("\nWhite wine describe()")
print(white_wine.describe())

print("\nAll columns describe()")
print(wine.describe(include='all'))

# 2. Numerical attributes
numeric_cols = wine.select_dtypes(include='number').columns.tolist()
print("\nNumerical columns:")
print(numeric_cols)

# 3. Missing values
print("\nMissing values in red wine:")
print(red_wine.isna().sum())
print("\nMissing values in white wine:")
print(white_wine.isna().sum())

# 4. Pandas queries
highest_quality = wine[wine.quality == wine.quality.max()]
avg_alcohol = wine.alcohol.mean()
high_alcohol = wine[wine.alcohol > avg_alcohol]
avg_quality_red = red_wine.quality.mean()
avg_quality_white = white_wine.quality.mean()
top10_quality = wine.sort_values(by=['quality', 'alcohol'], ascending=[False, False]).head(10)

print("\nHighest-quality wines:")
print(highest_quality)

print("\nAlcohol average:", avg_alcohol)
print("\nWines with alcohol above average:")
print(high_alcohol)

print("\nAverage quality red:", avg_quality_red)
print("Average quality white:", avg_quality_white)

print("\nTop 10 wines by quality:")
print(top10_quality)

# 5. Plots
plt.figure(figsize=(10, 6))
sns.histplot(wine.quality, bins=10, kde=False, color='purple')
plt.title("Histogram of Wine Quality")
plt.xlabel("Quality")
plt.ylabel("Count")
plt.show()

plt.figure(figsize=(6, 5))
plt.bar(['red', 'white'], [avg_quality_red, avg_quality_white], color=['darkred', 'gold'])
plt.title("Average Quality: Red vs White Wine")
plt.ylabel("Average Quality")
plt.show()

plt.figure(figsize=(8, 6))
sns.scatterplot(data=wine, x='alcohol', y='quality', hue='type', alpha=0.6)
plt.title("Alcohol vs Quality")
plt.xlabel("Alcohol")
plt.ylabel("Quality")
plt.show()

plt.figure(figsize=(12, 6))
sns.boxplot(data=wine, x='quality', y='alcohol')
plt.title("Alcohol Content by Quality Level")
plt.xlabel("Quality")
plt.ylabel("Alcohol")
plt.show()
```
