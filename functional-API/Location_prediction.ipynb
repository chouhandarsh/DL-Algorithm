```python
import tensorflow
from tensorflow import keras
from keras.layers import *
```

```python
from tensorflow.keras.models import Model
```

```python
X = Input(shape=(3,))
```

```python
hidden1 = Dense(128,activation='relu')(X) # this means that this is connected to X
hidden2 =  Dense(64,activation='relu')(hidden1) # this means that this is conncted  to hidden1 so we will have to name the node
output1=Dense(1,activation='linear')(hidden2)
output2 = Dense(1,activation='sigmoid')(hidden2)
```

```python
model=Model(inputs=X,outputs=[output1,output2])
```

```python
model.summary()
```

```python
from keras.utils import plot_model
plot_model(model,show_layer_names=True,show_shapes=4)
```
