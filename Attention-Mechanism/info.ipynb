# Attention Mechanism — Theory Notes

## Why Attention?
In a vanilla Seq2Seq (encoder-decoder) model, the encoder compresses the entire input sequence into a single fixed-length context vector (usually the final hidden state), which the decoder then uses to generate the entire output.

This creates a **bottleneck**: for long sequences, a single vector can't hold all the information, so translation/generation quality degrades as input length grows.

## Core Idea
Attention removes this bottleneck by letting the decoder look back at **all** encoder hidden states at every decoding step, instead of relying on just the last one.

At each output step, the model learns to assign different **attention weights** to each input word, focusing more on the parts of the input relevant to the word currently being generated.

## Encoder-Decoder Recap
- **Encoder**: processes input sequence (e.g., via RNN/LSTM/GRU) and produces a hidden state at each time step, not just the last one.
- **Decoder**: generates the output sequence one token at a time, using previous outputs and context.

## How Attention Works (Workflow)
1. Encoder produces hidden states `h1, h2, ..., hT` for each input word.
2. At each decoder time step `t`, compute an **alignment score** between the decoder's current state and each encoder hidden state — this measures relevance.
3. Apply **softmax** on these scores to get attention weights (they sum to 1).
4. Compute a **weighted sum** of encoder hidden states using these weights → this is the **context vector** for that step (dynamic, not fixed).
5. Combine this context vector with the decoder's current hidden state to predict the next output word.
6. Repeat for every decoding step, so the context vector changes dynamically depending on what's being generated.

## Key Takeaway
Attention lets the model "attend" to different input words dynamically for each output word, similar to how humans focus on relevant parts of a sentence while translating or summarizing. This significantly improves performance on long sequences and forms the foundation for later architectures (e.g., Transformers, which generalize this idea via self-attention).
