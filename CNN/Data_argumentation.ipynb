```python
import tensorflow as tf
from tensorflow import keras
from keras.preprocessing import image
from tensorflow.keras.preprocessing.image import ImageDataGenerator
```

```python
img = image.load_img("/run/media/darshchouhan/New Volume/DL-Algorithm/CNN/image copy 2.png")
```

```python
img
```

```python
datagen = ImageDataGenerator(
    rotation_range=40,
    shear_range=0.3,
    zoom_range=0.2,
    horizontal_flip = True,
    vertical_flip=False,
    width_shift_range=0.3,
    height_shift_range=0.4
)
```

```python
img = image.img_to_array(img)
```

```python
type(img)
```

```python
img.shape
```

```python
image_batch = img.reshape(1,333,500,3)
```

```python
i=0
for output in datagen.flow(image_batch,batch_size=1,save_to_dir="img"):
    i=i+1
    if i==20:
        break
```
