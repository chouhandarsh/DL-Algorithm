# Why RNNs?

## 1. Sequence → Any Length

- A **sequence** can have **any length**.
- A traditional **ANN** expects a **fixed input size**.
- Therefore:

  **Sequence → Any Length → ANN → Fixed Input Size**

- To make sequences compatible with an ANN, we can use **zero padding**.
- However:

  **Zero Padding → Cost of Computation ↑**

- Instead of padding, an **RNN** can process the sequence step-by-step and handle sequences of different lengths.
