# Types of RNN Architectures

RNN architectures are classified based on the shape of their **input** and **output** sequences. Before diving into Backpropagation Through Time, it's essential to understand these four types, since the backprop mechanics vary slightly across them.

There are **3 core types**, with one of them (Many-to-Many) having 2 sub-variants — so effectively **4 architectures**:

1. Many-to-One
2. One-to-Many
3. Many-to-Many (Same Length)
4. Many-to-Many (Variable Length)

---

## 1. Many-to-One

**Input:** Sequential data | **Output:** Single value (scalar)

The network reads a sequence step-by-step, and only produces a final output at the last time step.

### How it works
- At each time step, input is fed in along with the hidden state from the previous step
- The hidden state keeps getting updated and passed forward
- Only at the **final time step** is the actual output computed

### Applications
| Use Case | Input | Output |
|---|---|---|
| **Sentiment Analysis** | A sentence/text | 0 or 1 (negative/positive) |
| **Rating Prediction** | A movie review (text) | A star rating (1–5) |

```
x1 → [RNN] → h1 →
x2 → [RNN] → h2 →
x3 → [RNN] → h3 →
x4 → [RNN] → output (final prediction)
```

---

## 2. One-to-Many

**Input:** Single, non-sequential data | **Output:** Sequential data

This is the exact opposite of Many-to-One. A single input (like an image or a number) is fed once, and the network generates a sequence of outputs, one per time step — each output also feeds into the next step.

### Applications
| Use Case | Input | Output |
|---|---|---|
| **Image Captioning** | A single image | A sentence describing it (e.g. "A man is playing cricket") |
| **Music Generation** | A single seed/starting input | A continuous sequence of musical notes |

```
        ┌──────────────────────────────┐
image → [RNN] → y1 → [RNN] → y2 → [RNN] → y3 → ...
```

---

## 3. Many-to-Many (Same Length / Seq2Seq)

**Input:** Sequential data | **Output:** Sequential data (same length as input)

Also called **Sequence-to-Sequence**. Here, both input and output are sequences, and this type splits into two variants based on whether input length equals output length.

### 3a. Same Length
Input and output sequences have the **exact same number of elements**, and outputs are produced *simultaneously* at each time step, alongside the input.

**Applications:**
- **Part-of-Speech (POS) Tagging** — tag every word in a sentence as noun, verb, etc.
- **Named Entity Recognition (NER)** — for every word, decide whether it's an entity or not (e.g., "Let's meet at 7 PM at the airport")

```
x1 → [RNN] → y1
x2 → [RNN] → y2
x3 → [RNN] → y3
```
Each output is produced immediately as its corresponding input arrives.

---

## 4. Many-to-Many (Variable Length / Encoder–Decoder)

**Input:** Sequential data | **Output:** Sequential data (length can differ from input)

Unlike the same-length variant, here the model **first reads the entire input sequence completely**, and only *after* the last input is received does it start generating outputs.

### Why wait for the full input?
Tasks like translation require understanding the **full context, grammar, and meaning** of a sentence — not just word-by-word conversion. Translating word-by-word (like Same-Length M2M does) risks losing grammatical meaning and context. So this architecture waits until the whole input is read before producing any output.

### Applications
- **Machine Translation** — e.g., "Hey, my name is Nitish" → "मेरा नाम नीतीश है"
  (Note: the number of words in input vs. output language can differ — some languages express the same meaning more/less efficiently)

```
Input phase (Encoder):        Output phase (Decoder):
x1 → [RNN] →
x2 → [RNN] →
x3 → [RNN] →
x4 → [RNN] → (last input) → y1 → [RNN] → y2 → [RNN] → y3 → ...
```

> 💡 This is exactly why this architecture is also called an **Encoder–Decoder** network:
> - The part that consumes the input sequence = **Encoder**
> - The part that generates the output sequence = **Decoder**

---

## Bonus: One-to-One (Not Technically an RNN)

**Input:** Non-sequential | **Output:** Non-sequential

Example: **Image Classification** (image in → "cat" or "not cat" out). There's no recurrence, no feedback loop, and no multiple time steps — everything happens in a single pass. This is technically just a standard **ANN/CNN**, not an RNN, but it's useful to know so you can distinguish what does and doesn't count as recurrent.

## Summary Table

| Type | Input | Output | Example |
|---|---|---|---|
| One-to-One | Single | Single | Image classification (not really RNN) |
| One-to-Many | Single | Sequence | Image captioning, music generation |
| Many-to-One | Sequence | Single | Sentiment analysis, rating prediction |
| Many-to-Many (Same Length) | Sequence | Sequence (equal length) | POS tagging, NER |
| Many-to-Many (Variable Length) | Sequence | Sequence (different length) | Machine translation |
