How Embeddings Are Generated and How Relationships Are Established

1. Raw Text

We start with normal text:

"The cat is running"
"The dog is running"

The neural network cannot directly understand words, so we first convert the words into token IDs.

2. Tokenization

The tokenizer assigns an ID to every word/token.

For example:

cat → 10
dog → 25
running → 37
is → 5

So:

"The cat is running"
        ↓
[1, 10, 5, 37]

Important: Token IDs are arbitrary numbers.

There is no semantic relationship between:

cat → 10
dog → 25

The fact that 25 > 10 does NOT mean anything.

The token ID is simply an index/address used to find the corresponding vector.

3. Embedding Layer

Now we use:

Embedding(vocab_size, embedding_dimension)

For example:

Embedding(10000, 3)

This creates a trainable embedding matrix:

10000 words × 3 dimensions

Conceptually:

                 Embedding Matrix

Token ID       Vector
-----------------------------------------
0          →   [0.21, 0.73, -0.12]
1          →   [0.41, 0.22,  0.51]
2          →   [0.72, -0.11, 0.34]
...
10         →   [0.15, 0.81,  0.42]   ← cat
...
25         →   [-0.72, 0.11, 0.93]   ← dog
...
37         →   [0.31, 0.62,  0.17]   ← running

The embedding layer performs a lookup.

For example:

cat → 10

The embedding layer looks at:

Row 10

and returns:

[0.15, 0.81, 0.42]

Therefore:

Token ID
   ↓
Embedding Matrix
   ↓
Corresponding Vector

4. Initially, the Vectors Are Random

At the beginning of training, the embedding vectors do not contain meaningful relationships.

For example:

cat → [0.15, 0.81, 0.42]

dog → [-0.72, 0.11, 0.93]

The model does not initially know:

cat ≈ dog

The relationship has to be learned.

5. Embeddings Are Passed to the Neural Network

Suppose we have:

"The cat is running"

After tokenization:

[1, 10, 5, 37]

After embedding:

[
 [vector for "The"],
 [vector for "cat"],
 [vector for "is"],
 [vector for "running"]
]

These vectors are then passed into the architecture.

For example:

Token IDs
    ↓
Embedding
    ↓
Word Vectors
    ↓
RNN / LSTM / GRU
    ↓
Dense Layer
    ↓
Prediction

6. Prediction

Suppose we are doing sentiment classification.

Training data:

"I love cats" → Positive
"I love dogs" → Positive

"I hate cats" → Negative
"I hate dogs" → Negative

The model processes the embeddings and produces a prediction.

For example:

"I love cats"

Correct label:
1

Model prediction:
0.40

The prediction is incorrect.

7. Calculate the Loss

The model compares:

Actual value     = 1
Predicted value  = 0.40

and calculates the loss.

Prediction
    ↓
Compare with actual label
    ↓
Loss

The loss tells the model:

"How wrong was your prediction?"

8. Backpropagation

Now the error is propagated backward through the network.

Loss
 ↓
Dense Layer
 ↓
RNN
 ↓
Embedding Layer

The embedding vectors that contributed to the prediction receive gradients.

For example:

cat → [0.15, 0.81, 0.42]

may receive a gradient that causes it to change slightly:

cat → [0.18, 0.84, 0.40]

Similarly, the vector for dog can also be updated when dog participates in a training example.

9. This Happens Again and Again

The model sees many examples:

"The cat is running" → Animal
"The dog is running" → Animal

"The cat is eating" → Animal
"The dog is eating" → Animal

"The cat is sleeping" → Animal
"The dog is sleeping" → Animal

Notice the pattern:

cat → running
dog → running

cat → eating
dog → eating

cat → sleeping
dog → sleeping

The two words appear in similar contexts.

Therefore, during training, their embeddings can receive similar learning signals.

10. Similar Learning Signals → Similar Embeddings

The process is:

Similar contexts
       ↓
Similar effect on prediction
       ↓
Similar gradients
       ↓
Similar parameter updates
       ↓
Embeddings become similar

Eventually we might get something like:

cat → [0.71, 0.52, 0.21]

dog → [0.69, 0.55, 0.24]

Now the vectors are close in the embedding space.

Therefore:

cat ≈ dog

11. Where Does the Relationship Actually Come From?

The relationship does not come from the token IDs.

It does NOT come from:

cat → 10
dog → 25

Instead, it comes from the training process.

                TRAINING DATA
                     ↓
              Context / Usage
                     ↓
             Neural Network
                     ↓
                Prediction
                     ↓
                   Loss
                     ↓
             Backpropagation
                     ↓
            Gradient Updates
                     ↓
          Embedding Matrix Changes
                     ↓
        Similar words can move closer
                     ↓
          Learned relationships

12. Complete Flow

The entire process is:

                    RAW TEXT
                       ↓
                "The cat runs"
                       ↓
                  TOKENIZATION
                       ↓
                [1, 10, 37]
                       ↓
                    PADDING
                       ↓
              [1, 10, 37, 0, 0]
                       ↓
                  EMBEDDING
                       ↓
        ┌──────────────────────────┐
        │     Embedding Matrix     │
        │                          │
        │  ID 10 → [0.15, 0.81...] │
        │  ID 37 → [0.31, 0.62...] │
        └──────────────────────────┘
                       ↓
                 WORD VECTORS
                       ↓
                  RNN / LSTM
                       ↓
                    DENSE
                       ↓
                  PREDICTION
                       ↓
                     LOSS
                       ↓
                BACKPROPAGATION
                       ↓
              UPDATE PARAMETERS
                       ↓
        ┌──────────────────────────┐
        │  Embedding vectors also  │
        │       get updated        │
        └──────────────────────────┘
                       ↓
                 REPEAT MANY TIMES
                       ↓
             LEARNED EMBEDDINGS
                       ↓
        Similar contexts can produce
        similar vector representations

13. The Most Important Concept

Token ID

cat → 10

Means:

"Use row 10."

It has no semantic meaning.

Embedding

10 → [0.15, 0.81, 0.42]

Means:

"This is the current learned representation of cat."

Training

Prediction
    ↓
Loss
    ↓
Backpropagation
    ↓
Update Embedding Matrix

Means:

"Change the representation so that the model becomes better at its task."

14. One-Line Summary

Tokenization assigns an arbitrary ID to a word, the Embedding layer uses that ID to look up a trainable vector, and backpropagation continuously updates those vectors based on the training objective. Words that behave similarly in the training data can therefore develop similar representations.

Token ID
   ↓
"Which row?"
   ↓
Embedding Matrix
   ↓
"Which vector?"
   ↓
Neural Network
   ↓
Prediction
   ↓
Loss
   ↓
Backpropagation
   ↓
Update Embeddings
   ↓
Learn Relationships
