```python
with open(path,"r",encoding='utf8') as f:
    text=f.read()
print(len(text))
print(text[:200])
```

```python
import re
```

```python
def tokenize_text(text):

    text = text.lower()

    tokens = re.findall(
        r"\w+|[^\w\s]",
        text,
        re.UNICODE
    )

    return tokens
```

```python
tokens = tokenize_text(text)
```

```python
len(tokens)
```

```python
MAX_VOCAB_SIZE=12000
```

```python
from collections import Counter

counter = Counter(tokens)

print("\nUnique tokens:", len(counter))
```

```python
# most_common has word and idx correponding to the orginal word
```

```python
# we are reservinig 0 for the pad and 1  for the unknown
# now  we will extract the most common k characters for  it
most_common = counter.most_common(MAX_VOCAB_SIZE-2)
word_to_id = {
    "<PAD>":0,
    "<UNK>":1
}
for idx,(word,_) in enumerate(most_common,start=2):
    word_to_id[word]=idx
id_to_word={}

    
```

```python
for word,idx in word_to_id.items():
    id_to_word[idx] = word
```

```python
vocab_size = len(word_to_id)
```
