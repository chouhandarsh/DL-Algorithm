# Momentum

## Mathematical Intuition

The standard Gradient Descent update is:

$$
w_{t+1} = w_t - \eta \nabla w_t
$$

where:

- $\eta$ → learning rate
- $\nabla w_t$ → gradient at the current point

---

## Momentum

Instead of directly using only the current gradient, Momentum keeps track of the **past velocity**.

The velocity is defined as:

$$
\boxed{
v_t = \beta v_{t-1} + \eta \nabla w_t
}
$$

where:

- $\beta$ → decay factor
- $v_t$ → velocity at time $t$
- $\eta$ → learning rate
- $\nabla w_t$ → gradient at the current point

The weight update becomes:

$$
\boxed{
w_{t+1} = w_t - v_t
}
$$

---

## Combining the Two Equations

We have:

$$
v_t = \beta v_{t-1} + \eta \nabla w_t
$$

and:

$$
w_{t+1} = w_t - v_t
$$

Substituting $v_t$:

$$
w_{t+1}
=
w_t -
\left(
\beta v_{t-1}
+
\eta \nabla w_t
\right)
$$

Therefore:

$$
\boxed{
w_{t+1}
=
w_t -
\left(
\beta v_{t-1}
+
\eta \nabla w_t
\right)
}
$$

Here:

$$
\beta v_{t-1}
\quad \rightarrow \quad
\text{past velocity}
$$

and

$$
\eta \nabla w_t
\quad \rightarrow \quad
\text{gradient at the current point}
$$

---

## Intuition

The term

$$
\beta v_{t-1}
$$

allows the optimizer to **remember the direction of previous updates**.

The term

$$
\eta \nabla w_t
$$

represents the contribution from the **current gradient**.

Therefore, Momentum combines:

$$
\boxed{
\text{Past velocity}
+
\text{Current gradient}
}
$$

to determine the next movement of the weights.

---

# Nesterov Accelerated Gradient

Momentum is then extended to **Nesterov Accelerated Gradient (NAG)**.

The main idea is to consider the gradient after looking ahead in the direction of the accumulated momentum.

$$
\boxed{
\text{Nesterov Accelerated Gradient}
}

## Nesterov Accelerated Gradient

In Nesterov Accelerated Gradient, we first look ahead in the direction of the previous velocity.

The look-ahead weight is:

$$
\boxed{
w_{la} = w_t - \beta v_{t-1}
}
$$

where:

- $w_{la}$ → look-ahead weight
- $w_t$ → current weight
- $\beta v_{t-1}$ → contribution from the previous velocity

---

The velocity is then calculated using the gradient at the look-ahead point:

$$
\boxed{
v_t = \beta v_{t-1} + \eta \nabla w_{la}
}
$$

where:

- $\beta$ → decay factor
- $v_{t-1}$ → previous velocity
- $\eta$ → learning rate
- $\nabla w_{la}$ → gradient evaluated at the look-ahead weight

---

Finally, the weights are updated using the newly calculated velocity:

$$
\boxed{
w_{t+1} = w_t - v_t
}
$$

---

### Combining the Equations

The look-ahead weight is:

$$
w_{la} = w_t - \beta v_{t-1}
$$

The velocity becomes:

$$
v_t = \beta v_{t-1} + \eta \nabla w_{la}
$$

Substituting the look-ahead weight:

$$
\boxed{
v_t
=
\beta v_{t-1}
+
\eta
\nabla
\left(
w_t-\beta v_{t-1}
\right)
}
$$

Therefore, the final update is:

$$
\boxed{
w_{t+1}
=
w_t-v_t
}
$$

### Key Idea

Unlike standard Momentum, Nesterov Accelerated Gradient calculates the gradient at the **look-ahead position**:

$$
\boxed{
w_{la}=w_t-\beta v_{t-1}
}
$$

So instead of asking:

> What is the gradient at the current position?

NAG asks:

> **What will the gradient be at the position where momentum is taking us?**

Then it uses that gradient to make the next update.
