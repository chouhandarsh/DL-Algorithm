```python
from sklearn.base import BaseEstimator, RegressorMixin
import numpy as np

class MomentumRegressor(BaseEstimator, RegressorMixin):
    """Drop-in replacement for sklearn models"""
    
    def __init__(self, beta=0.9, learning_rate=0.01, epochs=100):
        self.beta = beta
        self.learning_rate = learning_rate
        self.epochs = epochs
        self.w = None
        self.V = None
    
    def fit(self, X, y):
        """Fit the model"""
        self.w = np.zeros(X.shape[1])
        self.V = np.zeros(X.shape[1])
        
        for epoch in range(self.epochs):
            y_pred = X @ self.w
            eta = 2 * X.T @ (y_pred - y) / len(y)
            
            self.V = self.beta * self.V + eta
            self.w = self.w - self.learning_rate * self.V
        
        return self
    
    def predict(self, X):
        """Make predictions"""
        return X @ self.w


# Now works like any sklearn model!
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

model = MomentumRegressor(beta=0.9, learning_rate=0.01, epochs=100)
model.fit(X_train, y_train)

train_score = model.score(X_train, y_train)
test_score = model.score(X_test, y_test)

print(f"Train R²: {train_score:.4f}")
print(f"Test R²:  {test_score:.4f}")
```
