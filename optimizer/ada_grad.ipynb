# Batch Gradient Descent

Consider a simple linear regression model:

$$
\hat{y} = wx + b
$$

where:

- $x$ → input feature
- $w$ → weight
- $b$ → bias
- $\hat{y}$ → predicted value

---

## Loss Function

For a single training example, the squared error loss is:

$$
L = (y-\hat{y})^2
$$

We calculate the gradients using the chain rule.

### Gradient with respect to $w$

$$
\frac{\partial L}{\partial w}
=
\frac{\partial L}{\partial \hat{y}}
\frac{\partial \hat{y}}{\partial w}
$$

Since:

$$
\hat{y}=wx+b
$$

we have:

$$
\frac{\partial \hat{y}}{\partial w}=x
$$

Therefore:

$$
\boxed{
\frac{\partial L}{\partial w}
=
-2(y-\hat{y})x
}
$$

---

### Gradient with respect to $b$

Similarly:

$$
\frac{\partial L}{\partial b}
=
\frac{\partial L}{\partial \hat{y}}
\frac{\partial \hat{y}}{\partial b}
$$

Since:

$$
\frac{\partial \hat{y}}{\partial b}=1
$$

we get:

$$
\boxed{
\frac{\partial L}{\partial b}
=
-2(y-\hat{y})\times 1
}
$$

or:

$$
\boxed{
\frac{\partial L}{\partial b}
=
-2(y-\hat{y})
}
$$

---

# Batch Gradient Descent

In **Batch Gradient Descent (BGD)**, we use the **entire training dataset** to calculate the gradient before updating the parameters.

For example, if we have:

$$
100\text{ rows}
$$

then all 100 rows are used to calculate the gradient.

The basic procedure is:

$$
\text{for } i \text{ in epochs:}
$$

Calculate the gradients using the complete dataset:

$$
\frac{\partial L}{\partial w}
$$

and

$$
\frac{\partial L}{\partial b}
$$

Then update the parameters:

$$
\boxed{
w = w-\eta\frac{\partial L}{\partial w}
}
$$

$$
\boxed{
b = b-\eta\frac{\partial L}{\partial b}
}
$$

where $\eta$ is the **learning rate**.

---

$$

# AdaGrad

## Core Formula

$$
\boxed{
V_t = V_{t-1} + (\nabla w_t)^2
}
$$

where:

- $V_t$ → accumulated squared gradients up to time $t$
- $V_{t-1}$ → accumulated squared gradients from previous steps
- $\nabla w_t$ → gradient of the loss with respect to the weight at step $t$

---

## Weight Update

$$
\boxed{
w_{t+1}
=
w_t
-
\frac{\eta}{\sqrt{V_t+\epsilon}}
\nabla w_t
}
$$

where:

- $w_t$ → current weight
- $w_{t+1}$ → updated weight
- $\eta$ → learning rate
- $V_t$ → accumulated squared gradients
- $\epsilon$ → small positive number added for numerical stability
- $\nabla w_t$ → current gradient

---

## Learning Rate

The effective learning rate is:

$$
\boxed{
\eta_t
=
\frac{\eta}{\sqrt{V_t+\epsilon}}
}
$$
**we are adding the gradient at the end and it means that the now it has the inverse relation with the slope**

Therefore:

$$
\boxed{
w_{t+1}
=
w_t-\eta_t\nabla w_t
}
$$

---

## Why $\epsilon$?

$$
\epsilon \rightarrow \text{small positive number}
$$

It prevents division by zero when:

$$
V_t \approx 0
$$

Typically:

$$
\epsilon > 0
$$

---

## Sign of the Update

The update is:

$$
w_{t+1}
=
w_t
-
\eta_t\nabla w_t
$$

Therefore:

### If the gradient is positive

$$
\nabla w_t > 0
$$

then:

$$
-\eta_t\nabla w_t < 0
$$

so the weight decreases:

$$
w_{t+1}<w_t
$$

### If the gradient is negative

$$
\nabla w_t < 0
$$

then:

$$
-\eta_t\nabla w_t > 0
$$

so the weight increases:

$$
w_{t+1}>w_t
$$

---

## Initialization

At the beginning:

$$
\boxed{
V_1=0
}
$$

Then the accumulated squared gradients are updated at every step:

$$
V_t
=
V_{t-1}
+
(\nabla w_t)^2
$$
