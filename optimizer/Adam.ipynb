## Adam is called Adaptive movement Estimation


- It is the most powerful optimization technique    

# Adam Optimizer

Adam combines the ideas of **Momentum** and **RMSProp**.

The weight update is:

$$
\boxed{
w_{t+1}
=
w_t
-
\frac{\eta}{\sqrt{V_t+\epsilon}}
\hat{m}_t
}
$$

where:

- $w_t$ → current weight
- $w_{t+1}$ → updated weight
- $\eta$ → learning rate
- $\hat{m}_t$ → bias-corrected first moment
- $V_t$ → second moment
- $\epsilon$ → small value for numerical stability

---

## First Moment — Momentum

$$
\boxed{
m_t
=
\beta_1m_{t-1}
+
(1-\beta_1)\nabla w_t
}
$$

where:

- $m_t$ → exponentially weighted average of gradients
- $m_{t-1}$ → previous first moment
- $\beta_1$ → decay factor for momentum
- $\nabla w_t$ → current gradient

Typically:

$$
\boxed{\beta_1=0.9}
$$

This part comes from **Momentum**.

---

## Second Moment — RMSProp / AdaGrad

$$
\boxed{
V_t
=
\beta_2V_{t-1}
+
(1-\beta_2)(\nabla w_t)^2
}
$$

where:

- $V_t$ → exponentially weighted average of squared gradients
- $V_{t-1}$ → previous second moment
- $\beta_2$ → decay factor for squared gradients
- $(\nabla w_t)^2$ → squared current gradient

Typically:

$$
\boxed{\beta_2=0.99}
$$

This part comes from **RMSProp / AdaGrad**.

---

# Bias Correction

Since both $m_t$ and $V_t$ are initialized close to zero, they are biased toward zero during the initial iterations.

Therefore, Adam applies bias correction.

### Bias-corrected first moment

$$
\boxed{
\hat{m}_t
=
\frac{m_t}{1-\beta_1^t}
}
$$

### Bias-corrected second moment

$$
\boxed{
\hat{V}_t
=
\frac{V_t}{1-\beta_2^t}
}
$$

Therefore, the Adam update can be written as:

$$
\boxed{
w_{t+1}
=
w_t
-
\eta
\frac{\hat{m}_t}
{\sqrt{\hat{V}_t}+\epsilon}
}
$$

---

## Hyperparameters

$$
\boxed{
\beta_1=0.9
}
$$

→ controls the **momentum / first moment**

$$
\boxed{
\beta_2=0.99
}
$$

→ controls the **squared-gradient / second moment**

$$
\boxed{
\eta=0.001
}
$$

→ learning rate (commonly used default)

$$
\boxed{
\epsilon
}
$$

→ small value used for numerical stability

---

# Adam = Momentum + RMSProp

The overall idea is:

$$
\boxed{
\text{Adam}
=
\text{Momentum}
+
\text{RMSProp}
+
\text{Bias Correction}
}
$$

### Momentum

$$
m_t
=
\beta_1m_{t-1}
+
(1-\beta_1)\nabla w_t
$$

### RMSProp

$$
V_t
=
\beta_2V_{t-1}
+
(1-\beta_2)(\nabla w_t)^2
$$

### Final Update

$$
w_{t+1}
=
w_t
-
\eta
\frac{\hat{m}_t}
{\sqrt{\hat{V}_t}+\epsilon}
$$
