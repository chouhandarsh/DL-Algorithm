### It is called exponentially weighted moving average
- This we give more importance to the the current points then the previous points
- The weightage of the next points will be more than the previous points

$$
V_1 = \beta V_\text{t-1}+(1-\beta)\theta_t
$$

- Here the B means that we are considering 1/(n-1) days for calcuating mean in our cacuation

# Mathematical Intuition

The exponential moving average is given by:

$$
V_t = \beta V_{t-1} + (1-\beta)\theta_t
$$

We start with:

$$
V_0 = 0
$$

For $t=1$:

$$
V_1 = (1-\beta)\theta_1
$$

For $t=2$:

$$
\begin{aligned}
V_2
&= \beta V_1 + (1-\beta)\theta_2 \\
&= \beta(1-\beta)\theta_1 + (1-\beta)\theta_2
\end{aligned}
$$

For $t=3$:

$$
\begin{aligned}
V_3
&= \beta V_2 + (1-\beta)\theta_3 \\
&= \beta^2(1-\beta)\theta_1
+ \beta(1-\beta)\theta_2
+ (1-\beta)\theta_3
\end{aligned}
$$

For $t=4$:

$$
\begin{aligned}
V_4
&= \beta V_3 + (1-\beta)\theta_4 \\
&= \beta^3(1-\beta)\theta_1
+ \beta^2(1-\beta)\theta_2
+ \beta(1-\beta)\theta_3
+ (1-\beta)\theta_4
\end{aligned}
$$

Taking $(1-\beta)$ common:

$$
\begin{aligned}
V_4
&= (1-\beta)
\left[
\beta^3\theta_1
+ \beta^2\theta_2
+ \beta\theta_3
+ \theta_4
\right]
\end{aligned}
$$

Since:

$$
0 < \beta < 1
$$

we have:

$$
\beta^3 < \beta^2 < \beta
$$

Therefore, the contribution of older values decreases exponentially:

$$
\boxed{
\beta^3 < \beta^2 < \beta < 1
}
$$

So the recent values of $\theta$ have a greater contribution to $V_t$, while older values have progressively smaller contributions.
