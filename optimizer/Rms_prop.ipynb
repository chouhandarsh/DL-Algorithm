# RMSProp

RMSProp maintains an exponentially weighted moving average of the **squared gradients**.

## Moving Average of Squared Gradients

$$
\boxed{
V_t
=
\beta V_{t-1}
+
(1-\beta)(\nabla w_t)^2
}
$$

where:

- $V_t$ → moving average of squared gradients at time $t$
- $V_{t-1}$ → moving average from the previous step
- $\beta$ → decay factor
- $\nabla w_t$ → gradient at the current step
- $(\nabla w_t)^2$ → squared gradient

---

## Weight Update

The weight is updated using the normalized gradient:

$$
\boxed{
w_{t+1}
=
w_t
-
\frac{\eta}{\sqrt{V_t+\epsilon}}
\nabla w_t
}
$$

where:

- $w_t$ → current weight
- $w_{t+1}$ → updated weight
- $\eta$ → learning rate
- $V_t$ → exponentially weighted average of squared gradients
- $\epsilon$ → small positive value for numerical stability
- $\nabla w_t$ → current gradient

---

## Exponential Decaying Average

RMSProp uses an **exponentially decaying average** of previous squared gradients.

Typically:

$$
\boxed{
\beta = 0.95
}
$$

A larger $\beta$ means that more importance is given to the **old values**.

For example:

$$
V_t
=
0.95V_{t-1}
+
0.05(\nabla w_t)^2
$$

Therefore:

$$
\underbrace{0.95}_{\text{old information}}
\qquad
\underbrace{0.05}_{\text{current gradient}}
$$

So RMSProp gives more weight to the history while still incorporating the current gradient.
