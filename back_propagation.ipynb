```python
def initialize_parameters(layer_dims):
    np.random.seed(3)
    parameters = {}
    L = len(layer_dims)
    for l in range(1,L):
        parameters["W"+str(l)] = np.ones((layer_dims[l-1],layer_dims[l]))*0.1
        parameters['b'+str(l)] = np.zeros((layer_dims[l],1))
    return parameters
         



```

```python
initialize_parameters([2,2,1])
```

```python
def linear_forward(A_prev,w,b):
    return np.dot(w.T,A_prev)+b
```

```python
def L_layer_forward(X,parameters):
    A = X
    L  = len(parameters)//2
    # Number of layers in the nural network
    for l in range(1,L+1):
        A_prev = A
        Wl = parameters["W"+str(l)]
        bl = parameters['b'+str(l)]
        # so we are passing the input with the Weights and the biases
        A = linear_forward(A_prev,Wl,bl)
        # the thing which we gave as the input and the result of the correponsding input
    return A,A_prev
```

```python
X =df[['cgpa','profile_score']].values[0].reshape(2,1)
# Shape (no of features, no.training examples )
y  = df[['lpa']].values[0][0]
parameters = initialize_parameters([2,2,1])
y_hat,A1 = L_layer_forward(X,parameters)
```

```python
def update_parameters(parameters,gradients,lr=0.01):
    L = len(parameters)//2
    for  l in range(1,L+1):
        parameters["W"+str(1)]-=lr()
```

```python
def update_parameters(parameters,y,y_hat,A1,X):
  parameters['W2'][0][0] = parameters['W2'][0][0] + (0.001 * 2 * (y - y_hat)*A1[0][0])
  parameters['W2'][1][0] = parameters['W2'][1][0] + (0.001 * 2 * (y - y_hat)*A1[1][0])
  parameters['b2'][0][0] = parameters['W2'][1][0] + (0.001 * 2 * (y - y_hat))

  parameters['W1'][0][0] = parameters['W1'][0][0] + (0.001 * 2 * (y - y_hat)*parameters['W2'][0][0]*X[0][0])
  parameters['W1'][0][1] = parameters['W1'][0][1] + (0.001 * 2 * (y - y_hat)*parameters['W2'][0][0]*X[1][0])
  parameters['b1'][0][0] = parameters['b1'][0][0] + (0.001 * 2 * (y - y_hat)*parameters['W2'][0][0])

  parameters['W1'][1][0] = parameters['W1'][1][0] + (0.001 * 2 * (y - y_hat)*parameters['W2'][1][0]*X[0][0])
  parameters['W1'][1][1] = parameters['W1'][1][1] + (0.001 * 2 * (y - y_hat)*parameters['W2'][1][0]*X[1][0])
  parameters['b1'][1][0] = parameters['b1'][1][0] + (0.001 * 2 * (y - y_hat)*parameters['W2'][1][0])
```

```python
parameters = initialize_parameters([2,2,1])
epochs=1000
for i in range(epochs):
    Loss =[]
    for j in range(df.shape[0]):
        X = df.iloc[:,0:-1].values[j].reshape(2,1)
        y  = df.iloc[:,-1].values[j]
        y_hat,A1 = L_layer_forward(X,parameters)
        y_hat = y_hat[0][0]
        update_parameters(parameters,y,y_hat,A1,X)
        Loss.append((y-y_hat)**2)
        print('Epochs - ',i+1,'Loss -',np.array(Loss).mean())
        
        
```

```python
import tensorflow
from tensorflow import keras
from keras import Sequential
from keras.layers import Dense
```

```python
model = Sequential()
model.add(Dense(2,input_dim=2))
model.add(Dense(1))
```

```python
model.summary()
```

```python
import numpy as np
```

```python
model.get_weights()
```

```python
new_weights = [np.array([[0.1,0.1],
                        [0.1,0.1]],dtype=np.float32),
                        np.array([0.,0.],dtype = np.float32),
                        np.array([[0.1],[0.1]],dtype=np.float32),
                        np.array([0.],dtype= np.float32)

                        ]
```

```python
model.set_weights(new_weights)
```

```python
model.get_weights()
```

```python
optimizer = keras.optimizers.Adam(learning_rate=0.01)
model.compile(loss="mse",optimizer=optimizer)
```

```python
model.fit(df.iloc[:,0:-1].values,df.iloc[:,-1].values,epochs=100)
```
