VGG16 Architecture — Notes



1. What is VGG16?

VGG16 is a Convolutional Neural Network (CNN) architecture used mainly for image classification.

The important idea in the diagram is that the network can be divided into two major parts:

Convolutional Base → extracts useful features from the image.

Fully Connected (FC) Part → uses those extracted features to perform classification.

So, conceptually:

$$
\boxed{\text{Image} \rightarrow \text{Feature Extraction} \rightarrow \text{Classification}}
$$

2. Input Image

The image enters the network from the left.

For example, an image can be represented as:

$$
X \in \mathbb{R}^{H \times W \times C}
$$

where:

$H$ = image height

$W$ = image width

$C$ = number of channels

RGB image → $C=3$

The CNN does not immediately convert the image into a normal 1D vector.

Instead, the convolutional layers initially preserve the spatial structure of the image.

That is why the convolutional part is useful for understanding things such as:

edges

corners

textures

shapes

object parts

3. Convolutional Base

The left portion of the diagram is the convolutional base.

It contains repeated blocks of:

$$
\text{Convolution} \rightarrow \text{Convolution} \rightarrow \text{Pooling}
$$

For VGG16, the convolutional blocks are:

Block 1: Conv1-1 → Conv1-2 → Pooling

Block 2: Conv2-1 → Conv2-2 → Pooling

Block 3: Conv3-1 → Conv3-2 → Conv3-3 → Pooling

Block 4: Conv4-1 → Conv4-2 → Conv4-3 → Pooling

Block 5: Conv5-1 → Conv5-2 → Conv5-3 → Pooling

The convolutional base is responsible for feature extraction.

4. What does a Convolution Layer do?

A convolution layer applies learnable filters/kernels to the image or feature map.

For a simple convolution:

$$
Z_{i,j} = \sum_m \sum_n X_{i+m,j+n}K_{m,n} + b
$$

where:

$X$ = input feature map

$K$ = learnable filter/kernel

$b$ = bias

$Z$ = resulting feature map

Each filter learns to detect some pattern.

For example, early layers may learn:

$$
\text{edges, lines, simple textures}
$$

while deeper layers can learn:

$$
\text{eyes, wheels, faces, object parts, complex shapes}
$$

5. Why are there multiple Convolution Layers?

This is one of the most important ideas in CNNs.

Suppose the input is a picture of a car.

The network gradually builds more meaningful representations:

Image
  ↓
Edges
  ↓
Corners + simple textures
  ↓
Shapes
  ↓
Object parts
  ↓
Complex object representation
  ↓
Car

Therefore:

$$
\boxed{\text{Early layers → low-level features}}
$$

$$
\boxed{\text{Deep layers → high-level features}}
$$

6. Pooling Layer

After several convolution layers, VGG16 uses a pooling layer.

The purpose of pooling is mainly to reduce the spatial dimensions of the feature maps.

For example:

Before pooling:

224 × 224 × 64

        ↓ Pooling

112 × 112 × 64

The height and width decrease, while the number of channels can remain the same.

A common operation is Max Pooling.

For a $2\times2$ region:

$$
\begin{bmatrix}
1 & 5\
3 & 2
\end{bmatrix}
\rightarrow 5
$$

The maximum value is selected.

Pooling therefore helps:

reduce computation

reduce spatial size

retain strong features

provide some translation robustness

7. What happens as we go deeper?

There is an important trade-off happening throughout the convolutional base.

Spatial information decreases

The feature maps become smaller:

Large feature map
      ↓
Smaller feature map
      ↓
Even smaller feature map

Number of feature channels increases

The network learns more types of features:

64 feature maps
      ↓
128 feature maps
      ↓
256 feature maps
      ↓
512 feature maps

So approximately:

$$
\boxed{\text{Spatial resolution} \downarrow}
$$

while:

$$
\boxed{\text{Feature depth} \uparrow}
$$

This is an important characteristic of VGG-style CNNs.

8. The Convolutional Base as a Feature Extractor

After passing through all convolutional blocks, the image has been transformed from raw pixels into a rich feature representation.

Initially:

Raw image

becomes:

Edges
Textures
Shapes
Object parts
Complex features

Therefore we can think of the convolutional base as:

$$
\boxed{
\text{Image} \xrightarrow{\text{CNN}} \text{Feature Representation}
}
$$

This is why the diagram labels this portion as:

CONV_BASE

9. Fully Connected Part

After the convolutional feature extraction, the network moves toward the fully connected layers.

The extracted feature maps need to be converted into a representation that the classifier can use.

Conceptually:

Feature Maps
     ↓
Flatten / feature representation
     ↓
Dense Layer
     ↓
Dense Layer
     ↓
Dense Layer

A fully connected neuron computes:

$$
z = w^Tx+b
$$

and then an activation function can be applied:

$$
a = f(z)
$$

Every neuron in a dense layer can use information from the previous layer.

10. Feature Extraction vs Classification

This is the most important conceptual division in the diagram.

Convolutional Base

The convolutional base answers:

"What features are present in this image?"

For example:

edges
textures
shapes
eyes
wheels
fur
faces
etc.

Fully Connected Part

The classification part answers:

"Given these features, which class does this image belong to?"

For example:

Extracted features
       ↓
Classifier
       ↓
Car

Therefore:

$$
\boxed{
\text{Conv Base} = \text{Feature Extraction}
}
$$

$$
\boxed{
\text{FC Part} = \text{Classification}
}
$$

11. Output Layer

The final layer produces the class predictions.

In the original ImageNet setup, VGG16 was trained to classify 1000 classes.

So the final output can be represented as:

$$
\mathbf{z} =
[z_1,z_2,\ldots,z_{1000}]
$$
$$
Each value corresponds to one class.

For example, conceptually:

Class 1   → score
Class 2   → score
Class 3   → score
...
Class 1000 → score

The model then converts these scores into probabilities using Softmax:

\frac{e^{z_k}}
{\sum_{j=1}^{1000}e^{z_j}}
$$

The probabilities satisfy:

$$
\sum_{k=1}^{1000}P(y=k|x)=1
$$

The predicted class is:

$$
\boxed{
\hat y = \arg\max_k P(y=k|x)
}
$$

So if the output probabilities are:

Car       → 0.72
Dog       → 0.05
Cat       → 0.03
Bus       → 0.15
...

the model predicts:

Car

12. What does "1000 classes" mean?

The note at the top of the diagram says:

ImageNet → 1000 classes

This means the original VGG16 model was trained to distinguish between 1000 ImageNet categories.

For example, the classes include different types of:

animals

vehicles

objects

instruments

tools

etc.

The final output layer therefore has:

$$
\boxed{1000\text{ output neurons}}
$$

for the original ImageNet model.

13. What happens if we use VGG16 for our own problem?

This is extremely important in Transfer Learning.

Suppose our own problem has only:

Cat
Dog

We don't need the original 1000-class classifier.

We can keep the pretrained convolutional base:

Pretrained VGG16
       ↓
Convolutional Base
       ↓
Extract useful image features
       ↓
Replace original classifier
       ↓
Our own classifier
       ↓
2 classes

For example:

$$
\boxed{
1000\text{ classes} \rightarrow 2\text{ classes}
}


The convolutional base can be reused because features such as edges, textures, and shapes are useful across many image-classification tasks.

14. Freezing the Convolutional Base

In transfer learning, we often freeze the convolutional base initially.

That means:
$$
$$
\boxed{\text{Weights of Conv Base are not updated}}
$$

Only the new classification layers are trained.

Conceptually:

Input Image
     ↓
VGG16 Conv Base
     ↓
[ FROZEN ]
     ↓
Feature Representation
     ↓
New Dense Layers
     ↓
Trainable
     ↓
Prediction

This allows us to use VGG16 as a pretrained feature extractor.

16. Why VGG16 is called a CNN

VGG16 is called a Convolutional Neural Network because convolutional layers are the main feature-extraction mechanism.

Its basic idea is:

$$
\boxed{
\text{Pixels}
\rightarrow
\text{Features}
\rightarrow
\text{High-level Representation}
\rightarrow
\text{Classification}
}
$$
$$
17. Understanding the Diagram in One Sentence

The easiest way to remember the whole architecture is:

VGG16 first uses convolution and pooling layers to extract increasingly complex spatial features from an image, then uses fully connected layers to map those features to class predictions.

19. The Big Picture

Think of VGG16 like a human looking at an image:

Raw Image
   ↓
"I see edges"
   ↓
"I see shapes and textures"
   ↓
"I see parts of an object"
   ↓
"These parts look like a car"
   ↓
"Prediction = Car"

The CNN performs a similar process mathematically:

$$

$$
X \rightarrow \text{Convolution} \rightarrow \text{Pooling} \rightarrow \text{Deep Features} \rightarrow \text{Dense Layers} \rightarrow \text{Softmax} \rightarrow \hat{y}
$$
